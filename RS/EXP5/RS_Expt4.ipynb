```python
merged_df = pd.merge(ratings_df, movies_df, on='movieId')
```

```python
avg_rating_df = merged_df.groupby('movieId')['rating'].mean().reset_index()
avg_rating_df = avg_rating_df.rename(columns={'rating': 'avg_rating'})
print("Average Rating per Movie:")
avg_rating_df.head()
```

```python
movies_with_ratings_df = pd.merge(movies_df, avg_rating_df, on='movieId')

print("Movies DataFrame with Average Ratings:")
display(movies_with_ratings_df.head())
```

```python
constraints = {
    'min_avg_rating': 4.0,
    'genres': ['Romance'],
    'min_release_year': 2000
}

print(f"Defined Constraints: {constraints}")
```

```python
constraints2 = {
    'min_avg_rating': 3.0,
    'genres': ['Comedy'],
    'min_release_year': 1995
}

print(f"Defined Constraints: {constraints}")
```

```python
def filter_movies(movies_data, current_constraints):
    filtered_movies = movies_data.copy()

    filtered_movies['year'] = filtered_movies['title'].str.extract(r'\((\d{4})\)', expand=False)
    filtered_movies['year'] = pd.to_numeric(filtered_movies['year'], errors='coerce')

    if 'min_release_year' in current_constraints:
        filtered_movies = filtered_movies[filtered_movies['year'] >= current_constraints['min_release_year']]

    if 'min_avg_rating' in current_constraints:
        filtered_movies = filtered_movies[filtered_movies['avg_rating'] >= current_constraints['min_avg_rating']]

    if 'genres' in current_constraints and current_constraints['genres']:
        genre_pattern = '|'.join(current_constraints['genres'])
        filtered_movies = filtered_movies[filtered_movies['genres'].str.contains(genre_pattern, case=False, na=False)]

    result_columns = ['movieId', 'title', 'genres', 'avg_rating']
    return filtered_movies[result_columns]
```

```python
recommended_movies = filter_movies(movies_with_ratings_df, constraints)
```
